```python
# ==========================================
# Lab 3: Correlation Analysis (Final Code)
# ==========================================

# Step 1: Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr, spearmanr


# Step 2: Create Sample Dataset
data = {
    "Math_Score": [78, 85, 90, 88, 76, 95],
    "Study_Hours": [5, 7, 9, 8, 4, 10],
    "Sleep_Hours": [7, 6, 8, 7, 6, 8]
}

df = pd.DataFrame(data)

print("Dataset:\n")
print(df)


# Step 3: Check Data Types and Missing Values
print("\nData Information:\n")
print(df.info())

print("\nMissing Values:\n")
print(df.isnull().sum())


# Step 4: Compute Pearson Correlation Matrix
correlation_matrix = df.corr(method="pearson")

print("\nPearson Correlation Matrix:\n")
print(correlation_matrix)


# Step 5: Visualize Correlation using Heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()


# Step 6: Correlation Between Study Hours and Math Score

# Pearson Correlation
pearson_corr, pearson_p = pearsonr(df["Study_Hours"], df["Math_Score"])

print("\nPearson Correlation (Study Hours vs Math Score):")
print("Correlation:", round(pearson_corr, 3))
print("P-value:", pearson_p)

# Spearman Correlation
spearman_corr, spearman_p = spearmanr(df["Study_Hours"], df["Math_Score"])

print("\nSpearman Correlation (Study Hours vs Math Score):")
print("Correlation:", round(spearman_corr, 3))
print("P-value:", spearman_p)


# Step 7: Scatter Plot
plt.figure(figsize=(6, 4))
sns.scatterplot(x="Study_Hours", y="Math_Score", data=df)
plt.title("Study Hours vs Math Score")
plt.xlabel("Study Hours")
plt.ylabel("Math Score")
plt.tight_layout()
plt.show()


# Step 8: Interpretation
print("\nInterpretation:")
if pearson_corr > 0:
    print("There is a positive relationship between Study Hours and Math Score.")
else:
    print("There is a negative relationship between Study Hours and Math Score.")

if pearson_p < 0.05:
    print("The relationship is statistically significant (p < 0.05).")
else:
    print("The relationship is NOT statistically significant (p >= 0.05).")
```
